```python
import pandas as pd
import matplotlib.pyplot as plt

dubai = pd.read_csv("../data/processed/dubai_residential_clean.csv")
ajman = pd.read_csv("../data/processed/ajman_clean.csv")

print("Dubai:", dubai.shape)
print("Ajman:", ajman.shape)
```

```python
print("Dubai Price Summary:")
print(dubai["price_aed"].describe())

print("\nMedian Price:")
print(f"AED {dubai['price_aed'].median():,.0f}")

print("\nAverage Price:")
print(f"AED {dubai['price_aed'].mean():,.0f}")
```

```python
# Show the 10 cheapest transactions
print("10 Cheapest Transactions:")
display(
    dubai.nsmallest(10, "price_aed")[
        ["price_aed", "area_sqm", "area", "property_type", "bedrooms", "year"]
    ]
)

# Show the 10 most expensive transactions
print("\n10 Most Expensive Transactions:")
display(
    dubai.nlargest(10, "price_aed")[
        ["price_aed", "area_sqm", "area", "property_type", "bedrooms", "year"]
    ]
)
```

```python
# Calculate price per square meter for EDA
dubai["price_per_sqm"] = dubai["price_aed"] / dubai["area_sqm"]

print("Price per sqm summary:")
print(dubai["price_per_sqm"].describe())

print("\nPercentiles:")
print(
    dubai["price_per_sqm"].quantile(
        [0.001, 0.005, 0.01, 0.05, 0.50, 0.95, 0.99, 0.995, 0.999]
    )
)
```
